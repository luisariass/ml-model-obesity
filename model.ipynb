```python
try:
    header_df = pd.read_excel("C:/Users/luisarias/Documents/ml-model-obesity/data/ObesityDataSet.xlsx", header=None, nrows=1)
    column_names_str = header_df.iloc[0, 0]
    correct_column_names = [name.strip() for name in column_names_str.split(',')] # strip() para quitar espacios extra
    print(f"Encabezados encontrados ({len(correct_column_names)}): {correct_column_names}")
except Exception as e:
    print(f"Error al leer la fila de encabezados: {e}")
    correct_column_names = [] # Proceder sin encabezados o detener
```

```python
try:
    # Lee todo el archivo, pero indica que no hay encabezado (ya los leímos) y salta la primera fila.
    df_data = pd.read_excel("C:/Users/luisarias/Documents/ml-model-obesity/data/ObesityDataSet.xlsx", header=None, skiprows=1)

    # --- Paso 3: Dividir la columna de datos ---
    # Asume que todos los datos están concatenados con comas en la primera columna del archivo leído.
    if not df_data.empty:
        df = df_data[df_data.columns[0]].str.split(',', expand=True)

        # --- Paso 4: Asignar y Verificar los nombres de columna ---
        if correct_column_names and len(correct_column_names) == len(df.columns):
            df.columns = correct_column_names
            print("Nombres de columna asignados correctamente.")
            display(df.head())
        elif not correct_column_names:
             print("No se pudieron leer los encabezados. El DataFrame no tendrá nombres de columna significativos.")
             display(df.head())
        else:
            # ¡Importante! Aquí descubres la discrepancia
            print(f"¡Error de Mapeo!")
            print(f"  - Número de encabezados leídos: {len(correct_column_names)}")
            print(f"  - Número de columnas generadas al dividir los datos: {len(df.columns)}")
            print("  Revisa si el delimitador (,) es correcto o si hay filas con un número diferente de valores.")
            # Puedes decidir mostrar el df sin nombres correctos o detener el proceso
            df.columns = [f'col_{i}' for i in range(len(df.columns))] # Nombres genéricos
            display(df.head())

    else:
        print("No se encontraron datos después de la fila de encabezados.")

except Exception as e:

    print(f"Error al procesar los datos: {e}")
```
